### **index()**

#### **NOTE**: the last index is -1 and the first index is the -(lenght of the string) 

```python
output = my_sequence.index('S')
print(output)
output = my_sequence.index(' ',-15, -1 )
print(output)
```

### **[*start*, *stop*, *step*]** _Note: stop index is not included_

**Start**: is the starting index and it is included in the output  
__Stop__: is the end index but not included  
**Step**: how many index from the start should be include. if your start is at index 0 and step is 4, then included index will be 0+4, 4+4, and 8+4


```python
substring = my_sequence[0:6]
print(substring)
substring = my_sequence[0:13:4]
print(substring)
```

#### Multi line string initialization 

```python
multi_line_string = """ I will make it      
***** at all cost
I will be consistent 
"""
print(multi_line_string)
```

```python
multi_line_string_stripped = multi_line_string.strip()
print(multi_line_string_stripped)
```

#### lstrip() rstrip() 
These are identical for removing whitespaces from either starting left side of a string or ending right side of a string

```python
multi_line_left_strip = multi_line_string.lstrip()
print(multi_line_left_strip)
```

### ***rjust()*** and ***ljust()***
use ***rjust()*** to add padding, this could be a whitespace which is the default or another character passed as the function's argument, this shifts the string to the right  
use ***ljust()*** to add padding, this could be a whitespace which is the default or another character passed as the function's argument, this shifts the string to the left

```python
multi_line_string_left_padding = multi_line_string.rjust(100)
print(multi_line_string_left_padding)
```

```python
multi_line_string_left_padding = multi_line_string.ljust(100, '*') // note, second arguement must just be a single character and this shipts to the left as well 
print(multi_line_string_left_padding)
```

### **split()**  
used to split a string by certain charaters in the string leading to a array where each index contains a substring of the original string based on the string delimiter  
Note; the defualt arguement of the split function is a white space but you can chage this by passing any argurment you wish to be the delimiter in the string

```python
split_this = 'I love God'
print(split_this.split())
```

```python
directory = "~/Home/Desktop/Senior DEV"
print(directory.split('/'))
```

```python
print(multi_line_string.split())
convert_back_to_string = multi_line_string.split()
```

### **join()** 
This is the opposite of **split()** it basically takes an array/list of strings and converts them into a single string delimited by whatever is appended with the join function

```python
back_to_string = " ".join(convert_back_to_string).rjust(100,'?').strip()
print(back_to_string)
```
